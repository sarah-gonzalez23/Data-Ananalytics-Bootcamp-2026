```python
import pandas as pd

def clean_missing_values(df):
    
    num_cols = df.select_dtypes(include=['number']).columns
    df[num_cols] = df[num_cols].fillna(df[num_cols].mean())

   
    cat_cols = df.select_dtypes(include=['object', 'string']).columns
    for col in cat_cols:
        if df[col].isna().sum() > 0:
            mode_value = df[col].mode(dropna=True)[0]
            df[col] = df[col].fillna(mode_value)

    
    date_cols = df.select_dtypes(include=['datetime']).columns
    for col in date_cols:
        df[col] = df[col].ffill().bfill()

    return df
```

```python
products_df = clean_missing_values(products_df)
sales_df = clean_missing_values(sales_df)
inventory_df = clean_missing_values(inventory_df)
stores_df = clean_missing_values(stores_df)
customers_df = clean_missing_values(customers_df)
```
